# SafeFundus — Notebook 05: Swin-Tiny Backbone Baseline

## Purpose

This notebook trains **Swin-Tiny** as the third backbone-screening baseline for SafeFundus.

For a fair comparison with ResNet50 v1 and DenseNet121, the training protocol is kept the same:

- Seed: **42**
- Input size: **224 × 224**
- ImageNet pretrained backbone
- Weighted BCEWithLogitsLoss
- AdamW
- Learning rate: **1e-4**
- Weight decay: **1e-4**
- Dropout: **0.30**
- Maximum epochs: **35**
- Early stopping patience: **7**
- LR scheduler patience: **3**
- Early stopping metric: **validation mAP**
- Validation-optimized per-class thresholds
- Same 14 known diseases
- Same RFMiD train/validation/test construction

Only the **backbone architecture** changes.

## Frozen RFMiD protocol

- **Known diseases:** 14
- **Training:** Normal + Known-only
- **Validation:** Normal + Known-only
- **Familiar internal test:** Normal + Known-only
- **Unknown safety test:** Unknown-only
- **Intermediate diseases:** CRVO and RPEC excluded
- **Mixed known + unknown:** excluded from the main baseline comparison

## Main outputs

The notebook saves:

- best Swin-Tiny checkpoint
- training history
- validation-optimized thresholds
- familiar-test predictions
- unknown-only predictions
- overall and per-class metrics
- rare-to-normal baseline safety error
- figures
- one complete ZIP:

`03_SwinTiny_Backbone_Baseline.zip`


```python
# ============================================================
# 1. Imports
# ============================================================

from pathlib import Path
import os
import json
import random
import time
import warnings
import shutil

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image

from sklearn.metrics import (
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
)

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from torchvision.models import Swin_T_Weights
from tqdm.auto import tqdm

warnings.filterwarnings("ignore")

print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

# 2. Configuration

These settings match the official ResNet50 v1 screening protocol.


```python
# ============================================================
# 2. Configuration
# ============================================================

CONFIG = {
    "seed": 42,
    "image_size": 224,
    "batch_size": 32,
    "epochs": 35,
    "learning_rate": 1e-4,
    "weight_decay": 1e-4,
    "dropout": 0.30,
    "num_workers": 2,
    "early_stopping_patience": 7,
    "lr_patience": 3,
    "threshold_grid_start": 0.05,
    "threshold_grid_end": 0.95,
    "threshold_grid_step": 0.01,
    "model_name": "swin_t",
}

DATA_ROOT = Path(
    "/kaggle/input/datasets/ozlemhakdagli/"
    "retinal-fundus-multi-disease-image-dataset-rfmid"
)

TRAIN_DIR = DATA_ROOT / "Training_set"
VAL_DIR   = DATA_ROOT / "Validation_set"
TEST_DIR  = DATA_ROOT / "Test_set"

TRAIN_CSV = TRAIN_DIR / "RFMiD_Training_Labels.csv"
VAL_CSV   = VAL_DIR   / "RFMiD_Validation_Labels.csv"
TEST_CSV  = TEST_DIR  / "RFMiD_Testing_Labels.csv"

NOTEBOOK_STEM = "03_SwinTiny_Backbone_Baseline"
OUTPUT_DIR = Path("/kaggle/working") / NOTEBOOK_STEM

CHECKPOINT_DIR = OUTPUT_DIR / "checkpoints"
TABLE_DIR = OUTPUT_DIR / "tables"
FIG_DIR = OUTPUT_DIR / "figures"
PRED_DIR = OUTPUT_DIR / "predictions"

for p in [OUTPUT_DIR, CHECKPOINT_DIR, TABLE_DIR, FIG_DIR, PRED_DIR]:
    p.mkdir(parents=True, exist_ok=True)

print(json.dumps(CONFIG, indent=2))
print("\nOutput directory:", OUTPUT_DIR)
```

# 3. Reproducibility


```python
# ============================================================
# 3. Reproducibility
# ============================================================

def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)

    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

seed_everything(CONFIG["seed"])

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("DEVICE:", DEVICE)
```

# 4. Frozen Known / Unknown Disease Protocol


```python
# ============================================================
# 4. Frozen disease protocol
# ============================================================

KNOWN_CLASSES = [
    "DR", "MH", "ODC", "TSLN", "DN", "MYA", "ARMD",
    "BRVO", "ODP", "ODE", "LS", "RS", "CSR", "CRS"
]

INTERMEDIATE_CLASSES = ["CRVO", "RPEC"]

print("Known classes:", len(KNOWN_CLASSES))
print(KNOWN_CLASSES)
print("\nIntermediate classes:", INTERMEDIATE_CLASSES)
```

# 5. Load RFMiD Labels and Reconstruct the Frozen Image Groups


```python
# ============================================================
# 5. Load labels and construct groups
# ============================================================

train_df = pd.read_csv(TRAIN_CSV)
val_df = pd.read_csv(VAL_CSV)
test_df = pd.read_csv(TEST_CSV)

ID_COL = "ID"
RISK_COL = "Disease_Risk"

assert ID_COL in train_df.columns
assert RISK_COL in train_df.columns

DISEASE_COLS = [
    c for c in train_df.columns
    if c not in {ID_COL, RISK_COL}
]

UNKNOWN_CLASSES = [
    c for c in DISEASE_COLS
    if c not in KNOWN_CLASSES and c not in INTERMEDIATE_CLASSES
]

def assign_group(row):
    positive = {c for c in DISEASE_COLS if int(row[c]) == 1}

    if len(positive) == 0:
        return "Normal"

    if positive & set(INTERMEDIATE_CLASSES):
        return "Contains_Intermediate"

    has_known = bool(positive & set(KNOWN_CLASSES))
    has_unknown = bool(positive & set(UNKNOWN_CLASSES))

    if has_known and has_unknown:
        return "Mixed_Known_Unknown"
    if has_known:
        return "Known_Only"
    if has_unknown:
        return "Unknown_Only"

    return "Other"

for df in [train_df, val_df, test_df]:
    df["Group"] = df.apply(assign_group, axis=1)

def show_group_counts(df, name):
    print(f"\n{name}")
    display(
        df["Group"]
        .value_counts()
        .rename_axis("Group")
        .reset_index(name="Count")
    )

show_group_counts(train_df, "TRAIN")
show_group_counts(val_df, "VALIDATION")
show_group_counts(test_df, "TEST")
```

# 6. Create the Experimental Splits


```python
# ============================================================
# 6. Experimental dataframes
# ============================================================

train_exp = train_df[
    train_df["Group"].isin(["Normal", "Known_Only"])
].copy()

val_exp = val_df[
    val_df["Group"].isin(["Normal", "Known_Only"])
].copy()

test_familiar = test_df[
    test_df["Group"].isin(["Normal", "Known_Only"])
].copy()

test_unknown = test_df[
    test_df["Group"] == "Unknown_Only"
].copy()

test_mixed = test_df[
    test_df["Group"] == "Mixed_Known_Unknown"
].copy()

print("Training images       :", len(train_exp))
print("Validation images     :", len(val_exp))
print("Familiar test images  :", len(test_familiar))
print("Unknown-only test     :", len(test_unknown))
print("Mixed exploratory test:", len(test_mixed))

assert len(train_exp) == 1713
assert len(val_exp) == 560
assert len(test_unknown) == 32
assert len(test_mixed) == 24

print("\n✓ Frozen protocol verified.")
```

# 7. Verify Image Paths


```python
# ============================================================
# 7. Add image paths
# ============================================================

def normalize_id(x):
    return str(int(float(x)))

def add_image_paths(df, folder):
    out = df.copy()

    out["image_path"] = out[ID_COL].apply(
        lambda x: str(folder / f"{normalize_id(x)}.png")
    )

    missing = [
        p for p in out["image_path"]
        if not Path(p).exists()
    ]

    if missing:
        raise FileNotFoundError(
            f"{len(missing)} image files are missing. "
            f"Example: {missing[:3]}"
        )

    return out

train_exp = add_image_paths(train_exp, TRAIN_DIR)
val_exp = add_image_paths(val_exp, VAL_DIR)
test_familiar = add_image_paths(test_familiar, TEST_DIR)
test_unknown = add_image_paths(test_unknown, TEST_DIR)
test_mixed = add_image_paths(test_mixed, TEST_DIR)

print("All image paths verified.")
```

# 8. Preprocessing and Augmentation

This matches the ResNet50 v1 augmentation protocol.


```python
# ============================================================
# 8. Transforms
# ============================================================

weights = Swin_T_Weights.IMAGENET1K_V1
imagenet_mean = weights.transforms().mean
imagenet_std = weights.transforms().std

train_transform = transforms.Compose([
    transforms.Resize(
        (CONFIG["image_size"], CONFIG["image_size"])
    ),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=10),
    transforms.ColorJitter(
        brightness=0.10,
        contrast=0.10,
        saturation=0.08,
        hue=0.02,
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=imagenet_mean,
        std=imagenet_std,
    ),
])

eval_transform = transforms.Compose([
    transforms.Resize(
        (CONFIG["image_size"], CONFIG["image_size"])
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=imagenet_mean,
        std=imagenet_std,
    ),
])

print("ImageNet mean:", imagenet_mean)
print("ImageNet std :", imagenet_std)
```

# 9. PyTorch Dataset


```python
# ============================================================
# 9. Dataset
# ============================================================

class RFMiDDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.df = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        image = Image.open(
            row["image_path"]
        ).convert("RGB")

        if self.transform is not None:
            image = self.transform(image)

        target = torch.tensor(
            row[KNOWN_CLASSES]
            .astype(np.float32)
            .values,
            dtype=torch.float32,
        )

        return {
            "image": image,
            "target": target,
            "id": normalize_id(row[ID_COL]),
            "group": row["Group"],
        }

train_dataset = RFMiDDataset(train_exp, train_transform)
val_dataset = RFMiDDataset(val_exp, eval_transform)
test_familiar_dataset = RFMiDDataset(test_familiar, eval_transform)
test_unknown_dataset = RFMiDDataset(test_unknown, eval_transform)

print("Train dataset     :", len(train_dataset))
print("Validation dataset:", len(val_dataset))
print("Familiar test     :", len(test_familiar_dataset))
print("Unknown test      :", len(test_unknown_dataset))
```

# 10. DataLoaders


```python
# ============================================================
# 10. DataLoaders
# ============================================================

def seed_worker(worker_id):
    worker_seed = CONFIG["seed"] + worker_id
    np.random.seed(worker_seed)
    random.seed(worker_seed)

generator = torch.Generator()
generator.manual_seed(CONFIG["seed"])

train_loader = DataLoader(
    train_dataset,
    batch_size=CONFIG["batch_size"],
    shuffle=True,
    num_workers=CONFIG["num_workers"],
    pin_memory=True,
    persistent_workers=CONFIG["num_workers"] > 0,
    worker_init_fn=seed_worker,
    generator=generator,
)

val_loader = DataLoader(
    val_dataset,
    batch_size=CONFIG["batch_size"],
    shuffle=False,
    num_workers=CONFIG["num_workers"],
    pin_memory=True,
    persistent_workers=CONFIG["num_workers"] > 0,
)

test_familiar_loader = DataLoader(
    test_familiar_dataset,
    batch_size=CONFIG["batch_size"],
    shuffle=False,
    num_workers=CONFIG["num_workers"],
    pin_memory=True,
    persistent_workers=CONFIG["num_workers"] > 0,
)

test_unknown_loader = DataLoader(
    test_unknown_dataset,
    batch_size=CONFIG["batch_size"],
    shuffle=False,
    num_workers=CONFIG["num_workers"],
    pin_memory=True,
    persistent_workers=CONFIG["num_workers"] > 0,
)

batch = next(iter(train_loader))

print("Image batch shape :", batch["image"].shape)
print("Target batch shape:", batch["target"].shape)
```

# 11. Weighted BCE for Class Imbalance


```python
# ============================================================
# 11. Positive class weights
# ============================================================

positive_counts = train_exp[
    KNOWN_CLASSES
].sum(axis=0).astype(float)

negative_counts = (
    len(train_exp) - positive_counts
)

pos_weight = (
    negative_counts /
    np.maximum(positive_counts, 1.0)
)

pos_weight = np.minimum(
    pos_weight,
    20.0,
)

class_balance_df = pd.DataFrame({
    "Disease": KNOWN_CLASSES,
    "Positive_Train": positive_counts.values.astype(int),
    "Negative_Train": negative_counts.values.astype(int),
    "Pos_Weight": pos_weight.values,
})

display(class_balance_df)

class_balance_df.to_csv(
    TABLE_DIR / "swintiny_class_balance.csv",
    index=False,
)

POS_WEIGHT_TENSOR = torch.tensor(
    pos_weight.values,
    dtype=torch.float32,
    device=DEVICE,
)
```

# 12. Build Swin-Tiny

Torchvision Swin-Tiny uses a final `head` linear layer after the transformer features, normalization, global pooling, and flattening.

For the RFMiD multi-label task we replace that final head with:

Dropout(0.30) → Linear(features → 14)

The model still returns **raw logits**; sigmoid is applied only during evaluation.


```python
# ============================================================
# 12. Model
# ============================================================

def build_swin_tiny(
    num_classes,
    dropout=0.3,
):
    try:
        model = models.swin_t(
            weights=Swin_T_Weights.IMAGENET1K_V1
        )

        print(
            "Loaded ImageNet pretrained "
            "Swin-Tiny weights."
        )

    except Exception as e:
        print(
            "WARNING: Could not load "
            "pretrained weights."
        )
        print("Reason:", e)

        print(
            "\nFor the real experiment, enable Kaggle Internet "
            "or provide pretrained weights. Training from scratch "
            "is NOT recommended."
        )

        model = models.swin_t(
            weights=None
        )

    in_features = model.head.in_features

    model.head = nn.Sequential(
        nn.Dropout(p=dropout),
        nn.Linear(
            in_features,
            num_classes,
        ),
    )

    return model

model = build_swin_tiny(
    num_classes=len(KNOWN_CLASSES),
    dropout=CONFIG["dropout"],
).to(DEVICE)

print(model.head)
```

# 13. Loss, Optimizer, Scheduler


```python
# ============================================================
# 13. Training components
# ============================================================

criterion = nn.BCEWithLogitsLoss(
    pos_weight=POS_WEIGHT_TENSOR
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=CONFIG["learning_rate"],
    weight_decay=CONFIG["weight_decay"],
)

scheduler = (
    torch.optim.lr_scheduler
    .ReduceLROnPlateau(
        optimizer,
        mode="max",
        factor=0.5,
        patience=CONFIG["lr_patience"],
    )
)

scaler = torch.cuda.amp.GradScaler(
    enabled=(DEVICE.type == "cuda")
)

print("Loss      : BCEWithLogitsLoss + pos_weight")
print("Optimizer : AdamW")
print("Scheduler : ReduceLROnPlateau on validation mAP")
print("AMP       :", DEVICE.type == "cuda")
```

# 14. Metric Functions


```python
# ============================================================
# 14. Metrics
# ============================================================

def safe_macro_auroc(y_true, y_prob):
    scores = []

    for c in range(y_true.shape[1]):
        if len(np.unique(y_true[:, c])) < 2:
            continue

        scores.append(
            roc_auc_score(
                y_true[:, c],
                y_prob[:, c],
            )
        )

    return float(np.mean(scores)) if scores else np.nan

def safe_map(y_true, y_prob):
    scores = []

    for c in range(y_true.shape[1]):
        if y_true[:, c].sum() == 0:
            continue

        scores.append(
            average_precision_score(
                y_true[:, c],
                y_prob[:, c],
            )
        )

    return float(np.mean(scores)) if scores else np.nan

def compute_epoch_metrics(y_true, y_prob, threshold=0.5):
    y_pred = (y_prob >= threshold).astype(int)

    return {
        "macro_f1": f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0,
        ),
        "micro_f1": f1_score(
            y_true,
            y_pred,
            average="micro",
            zero_division=0,
        ),
        "mAP": safe_map(
            y_true,
            y_prob,
        ),
        "macro_auroc": safe_macro_auroc(
            y_true,
            y_prob,
        ),
    }
```

# 15. Training and Validation Functions


```python
# ============================================================
# 15. Train / validation functions
# ============================================================

def train_one_epoch(
    model,
    loader,
    optimizer,
    criterion,
    scaler,
):
    model.train()

    running_loss = 0.0
    all_targets = []
    all_probs = []

    pbar = tqdm(
        loader,
        desc="Train",
        leave=False,
    )

    for batch in pbar:
        images = batch["image"].to(
            DEVICE,
            non_blocking=True,
        )

        targets = batch["target"].to(
            DEVICE,
            non_blocking=True,
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        with torch.cuda.amp.autocast(
            enabled=(DEVICE.type == "cuda")
        ):
            logits = model(images)
            loss = criterion(
                logits,
                targets,
            )

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        probs = torch.sigmoid(logits)

        running_loss += (
            loss.item() *
            images.size(0)
        )

        all_targets.append(
            targets.detach()
            .cpu()
            .numpy()
        )

        all_probs.append(
            probs.detach()
            .cpu()
            .numpy()
        )

        pbar.set_postfix(
            loss=f"{loss.item():.4f}"
        )

    y_true = np.concatenate(all_targets)
    y_prob = np.concatenate(all_probs)

    metrics = compute_epoch_metrics(
        y_true,
        y_prob,
    )

    metrics["loss"] = (
        running_loss /
        len(loader.dataset)
    )

    return metrics


@torch.no_grad()
def evaluate_loader(
    model,
    loader,
    criterion=None,
):
    model.eval()

    running_loss = 0.0
    all_targets = []
    all_probs = []
    all_logits = []
    all_ids = []
    all_groups = []

    for batch in tqdm(
        loader,
        desc="Evaluate",
        leave=False,
    ):
        images = batch["image"].to(
            DEVICE,
            non_blocking=True,
        )

        targets = batch["target"].to(
            DEVICE,
            non_blocking=True,
        )

        with torch.cuda.amp.autocast(
            enabled=(DEVICE.type == "cuda")
        ):
            logits = model(images)

            if criterion is not None:
                loss = criterion(
                    logits,
                    targets,
                )

                running_loss += (
                    loss.item() *
                    images.size(0)
                )

        probs = torch.sigmoid(logits)

        all_targets.append(
            targets.cpu().numpy()
        )

        all_probs.append(
            probs.cpu().numpy()
        )

        all_logits.append(
            logits.cpu().numpy()
        )

        all_ids.extend(
            batch["id"]
        )

        all_groups.extend(
            batch["group"]
        )

    y_true = np.concatenate(all_targets)
    y_prob = np.concatenate(all_probs)
    logits = np.concatenate(all_logits)

    metrics = compute_epoch_metrics(
        y_true,
        y_prob,
    )

    if criterion is not None:
        metrics["loss"] = (
            running_loss /
            len(loader.dataset)
        )

    return {
        "metrics": metrics,
        "targets": y_true,
        "probs": y_prob,
        "logits": logits,
        "ids": all_ids,
        "groups": all_groups,
    }
```

# 16. Train Swin-Tiny

The best checkpoint is selected by **validation mAP**.


```python
# ============================================================
# 16. Training
# ============================================================

history = []

best_val_map = -np.inf
epochs_without_improvement = 0

best_checkpoint_path = (
    CHECKPOINT_DIR /
    "best_swintiny_rfmid.pth"
)

start_time = time.time()

for epoch in range(
    1,
    CONFIG["epochs"] + 1,
):
    print(f"\n{'='*70}")
    print(
        f"Epoch {epoch}/"
        f"{CONFIG['epochs']}"
    )
    print(f"{'='*70}")

    train_metrics = train_one_epoch(
        model,
        train_loader,
        optimizer,
        criterion,
        scaler,
    )

    val_result = evaluate_loader(
        model,
        val_loader,
        criterion,
    )

    val_metrics = val_result["metrics"]

    scheduler.step(
        val_metrics["mAP"]
    )

    current_lr = optimizer.param_groups[0]["lr"]

    row = {
        "epoch": epoch,
        "lr": current_lr,

        "train_loss": train_metrics["loss"],
        "train_macro_f1": train_metrics["macro_f1"],
        "train_micro_f1": train_metrics["micro_f1"],
        "train_mAP": train_metrics["mAP"],
        "train_macro_auroc": train_metrics["macro_auroc"],

        "val_loss": val_metrics["loss"],
        "val_macro_f1": val_metrics["macro_f1"],
        "val_micro_f1": val_metrics["micro_f1"],
        "val_mAP": val_metrics["mAP"],
        "val_macro_auroc": val_metrics["macro_auroc"],
    }

    history.append(row)

    print(
        f"Train Loss={train_metrics['loss']:.4f} | "
        f"mAP={train_metrics['mAP']:.4f} | "
        f"Macro-F1={train_metrics['macro_f1']:.4f}"
    )

    print(
        f"Val   Loss={val_metrics['loss']:.4f} | "
        f"mAP={val_metrics['mAP']:.4f} | "
        f"Macro-F1={val_metrics['macro_f1']:.4f} | "
        f"AUROC={val_metrics['macro_auroc']:.4f}"
    )

    print(f"LR={current_lr:.2e}")

    if val_metrics["mAP"] > best_val_map:
        best_val_map = val_metrics["mAP"]
        epochs_without_improvement = 0

        torch.save({
            "model_state_dict": model.state_dict(),
            "config": CONFIG,
            "known_classes": KNOWN_CLASSES,
            "best_val_map": best_val_map,
            "epoch": epoch,
        }, best_checkpoint_path)

        print(
            "✓ New best checkpoint saved: "
            f"val mAP={best_val_map:.4f}"
        )
    else:
        epochs_without_improvement += 1

        print(
            "No validation mAP improvement for "
            f"{epochs_without_improvement} epoch(s)."
        )

    if (
        epochs_without_improvement >=
        CONFIG["early_stopping_patience"]
    ):
        print("\nEarly stopping triggered.")
        break

elapsed_minutes = (
    time.time() - start_time
) / 60

history_df = pd.DataFrame(history)

history_df.to_csv(
    TABLE_DIR /
    "swintiny_training_history.csv",
    index=False,
)

print(
    f"\nTraining finished in "
    f"{elapsed_minutes:.1f} minutes."
)

print(
    "Best validation mAP:",
    best_val_map,
)

print(
    "Best checkpoint:",
    best_checkpoint_path,
)

display(history_df.tail())
```

# 19. Optimize Per-Class Thresholds on Validation Data Only


```python
# ============================================================
# 19. Threshold optimization
# ============================================================

val_result = evaluate_loader(
    model,
    val_loader,
)

val_true = val_result["targets"]
val_prob = val_result["probs"]

threshold_grid = np.arange(
    CONFIG["threshold_grid_start"],
    CONFIG["threshold_grid_end"] + 1e-9,
    CONFIG["threshold_grid_step"],
)

def optimize_thresholds(
    y_true,
    y_prob,
):
    thresholds = []

    for c, disease in enumerate(
        KNOWN_CLASSES
    ):
        best_threshold = 0.5
        best_f1 = -1.0

        if y_true[:, c].sum() == 0:
            thresholds.append(
                best_threshold
            )
            continue

        for threshold in threshold_grid:
            pred = (
                y_prob[:, c] >= threshold
            ).astype(int)

            score = f1_score(
                y_true[:, c],
                pred,
                zero_division=0,
            )

            if score > best_f1:
                best_f1 = score
                best_threshold = float(
                    threshold
                )

        thresholds.append(
            best_threshold
        )

    return np.array(
        thresholds,
        dtype=np.float32,
    )

CLASS_THRESHOLDS = optimize_thresholds(
    val_true,
    val_prob,
)

threshold_df = pd.DataFrame({
    "Disease": KNOWN_CLASSES,
    "Threshold": CLASS_THRESHOLDS,
})

display(threshold_df)

threshold_df.to_csv(
    TABLE_DIR /
    "swintiny_validation_thresholds.csv",
    index=False,
)
```

# 20. Familiar-Test Evaluation


```python
# ============================================================
# 20. Familiar-test evaluation
# ============================================================

familiar_result = evaluate_loader(
    model,
    test_familiar_loader,
)

y_true = familiar_result["targets"]
y_prob = familiar_result["probs"]

y_pred = (
    y_prob >=
    CLASS_THRESHOLDS[None, :]
).astype(int)

overall_metrics = {
    "model": "Swin-Tiny",
    "seed": CONFIG["seed"],
    "familiar_test_images":
        int(len(y_true)),
    "macro_f1": float(
        f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0,
        )
    ),
    "micro_f1": float(
        f1_score(
            y_true,
            y_pred,
            average="micro",
            zero_division=0,
        )
    ),
    "mAP": safe_map(
        y_true,
        y_prob,
    ),
    "macro_auroc":
        safe_macro_auroc(
            y_true,
            y_prob,
        ),
}

print(
    json.dumps(
        overall_metrics,
        indent=2,
    )
)

with open(
    TABLE_DIR /
    "swintiny_overall_metrics.json",
    "w",
) as f:
    json.dump(
        overall_metrics,
        f,
        indent=2,
    )
```

# 21. Per-Disease Metrics


```python
# ============================================================
# 21. Per-class metrics
# ============================================================

rows = []

for c, disease in enumerate(
    KNOWN_CLASSES
):
    yt = y_true[:, c]
    yp = y_pred[:, c]
    prob = y_prob[:, c]

    tn, fp, fn, tp = (
        confusion_matrix(
            yt,
            yp,
            labels=[0, 1],
        ).ravel()
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )

    auroc = (
        roc_auc_score(
            yt,
            prob,
        )
        if len(np.unique(yt)) == 2
        else np.nan
    )

    ap = (
        average_precision_score(
            yt,
            prob,
        )
        if yt.sum() > 0
        else np.nan
    )

    rows.append({
        "Disease": disease,
        "Positive_Test":
            int(yt.sum()),
        "Threshold":
            float(
                CLASS_THRESHOLDS[c]
            ),
        "Precision":
            precision_score(
                yt,
                yp,
                zero_division=0,
            ),
        "Sensitivity_Recall":
            recall_score(
                yt,
                yp,
                zero_division=0,
            ),
        "Specificity":
            specificity,
        "F1":
            f1_score(
                yt,
                yp,
                zero_division=0,
            ),
        "AUROC":
            auroc,
        "Average_Precision":
            ap,
        "TP": int(tp),
        "FP": int(fp),
        "TN": int(tn),
        "FN": int(fn),
    })

per_class_metrics = pd.DataFrame(rows)

display(per_class_metrics)

per_class_metrics.to_csv(
    TABLE_DIR /
    "swintiny_per_class_metrics.csv",
    index=False,
)
```

# 22. Save Familiar-Test Predictions


```python
# ============================================================
# 22. Save familiar-test predictions
# ============================================================

pred_df = pd.DataFrame({
    "ID": familiar_result["ids"],
    "Group": familiar_result["groups"],
})

for c, disease in enumerate(
    KNOWN_CLASSES
):
    pred_df[
        f"true_{disease}"
    ] = y_true[:, c].astype(int)

    pred_df[
        f"logit_{disease}"
    ] = (
        familiar_result["logits"]
        [:, c]
    )

    pred_df[
        f"prob_{disease}"
    ] = y_prob[:, c]

    pred_df[
        f"pred_{disease}"
    ] = y_pred[:, c].astype(int)

pred_df.to_csv(
    PRED_DIR /
    "swintiny_familiar_test_predictions.csv",
    index=False,
)

display(pred_df.head())
```

# 23. Closed-Set Safety Test on Unknown Rare Diseases


```python
# ============================================================
# 23. Unknown-only safety evaluation
# ============================================================

unknown_result = evaluate_loader(
    model,
    test_unknown_loader,
)

unknown_prob = unknown_result["probs"]

unknown_pred = (
    unknown_prob >=
    CLASS_THRESHOLDS[None, :]
).astype(int)

unknown_predicted_normal = (
    unknown_pred.sum(axis=1) == 0
)

rare_to_normal_count = int(
    unknown_predicted_normal.sum()
)

rare_to_normal_rate = float(
    unknown_predicted_normal.mean()
)

unknown_forced_into_known_rate = float(
    (~unknown_predicted_normal).mean()
)

safety_metrics = {
    "model": "Swin-Tiny",
    "unknown_only_test_images":
        int(len(unknown_pred)),
    "rare_to_normal_count":
        rare_to_normal_count,
    "rare_to_normal_error_rate":
        rare_to_normal_rate,
    "unknown_forced_into_known_rate":
        unknown_forced_into_known_rate,
}

print(
    json.dumps(
        safety_metrics,
        indent=2,
    )
)

with open(
    TABLE_DIR /
    "swintiny_unknown_safety_metrics.json",
    "w",
) as f:
    json.dump(
        safety_metrics,
        f,
        indent=2,
    )
```
